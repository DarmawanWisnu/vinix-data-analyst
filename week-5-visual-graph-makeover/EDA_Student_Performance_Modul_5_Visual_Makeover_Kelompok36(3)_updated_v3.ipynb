## 4. Profil Statistik

- **Numerik:** mean, median, modus, simpangan baku, minimum, maksimum.
- **Kategorikal:** jumlah kategori, modus, frekuensi, dan proporsi.


```python

# Kolom numerik
numeric_cols = dataset.select_dtypes(include=np.number).columns.tolist()

numeric_profile = pd.DataFrame({
    "Mean": dataset[numeric_cols].mean(),
    "Median": dataset[numeric_cols].median(),
    "Mode": [dataset[col].mode().iloc[0] for col in numeric_cols],
    "Std Dev": dataset[numeric_cols].std(),
    "Minimum": dataset[numeric_cols].min(),
    "Maximum": dataset[numeric_cols].max()
}).round(2)

print("STATISTICAL PROFILING - NUMERIK")
display(numeric_profile)
```

```python

# Kolom kategorikal
categorical_cols = dataset.select_dtypes(include="object").columns.tolist()

categorical_rows = []
for col in categorical_cols:
    vc = dataset[col].value_counts()
    categorical_rows.append({
        "Column": col,
        "Unique Categories": dataset[col].nunique(),
        "Mode": dataset[col].mode().iloc[0],
        "Mode Frequency": int(vc.iloc[0]),
        "Mode Proportion (%)": round(vc.iloc[0] / len(dataset) * 100, 2)
    })

categorical_profile = pd.DataFrame(categorical_rows)

print("STATISTICAL PROFILING - KATEGORIKAL")
display(categorical_profile)

print("\nDistribusi Extracurricular Activities:")
display(
    dataset["Extracurricular Activities"]
    .value_counts()
    .rename_axis("Category")
    .to_frame("Count")
    .assign(Proportion=lambda x: (x["Count"] / len(dataset) * 100).round(2))
)
```

**Hipotesis:** Bagi kelompok nilai awal rendah, program yang menggabungkan remedial konsep dan pembinaan kebiasaan belajar berpotensi lebih tepat daripada program jam belajar saja. Kelompok nilai awal menengah dengan jam belajar rendah dapat menjadi sasaran dukungan konsistensi belajar.

## 8. Analisis Tambahan: Model Prediksi

**Linear Regression** digunakan untuk memperkirakan Performance Index dari variabel yang tersedia. Target berupa angka kontinu dan pola pada EDA tampak mendekati linear. Koefisien model membantu menjelaskan asosiasi tiap variabel setelah variabel lain diperhitungkan.


```python

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Untuk evaluasi model, exact duplicate dihapus agar tidak mudah muncul di train dan test sekaligus
dataset_model = dataset.drop_duplicates().copy()

X = dataset_model.drop(columns="Performance Index")
y = dataset_model["Performance Index"]

categorical_features = ["Extracurricular Activities"]
numeric_features = [col for col in X.columns if col not in categorical_features]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numeric_features),
        ("cat", OneHotEncoder(drop="first", handle_unknown="ignore"), categorical_features)
    ]
)

model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print("MODEL PERFORMANCE")
print(f"Data model setelah drop exact duplicate: {len(dataset_model):,} rows")
print(f"Train rows: {len(X_train):,}")
print(f"Test rows : {len(X_test):,}")
print(f"R²        : {r2:.4f}")
print(f"MAE       : {mae:.4f}")
print(f"RMSE      : {rmse:.4f}")
```

```python

# Mengambil koefisien model agar hasil prediksi tetap interpretable
ohe = model.named_steps["preprocessor"].named_transformers_["cat"]
encoded_names = ohe.get_feature_names_out(categorical_features).tolist()
feature_names = numeric_features + encoded_names

coefficients = model.named_steps["regressor"].coef_

coef_df = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
}).sort_values("Coefficient", key=np.abs, ascending=False)

print("KOEFISIEN LINEAR REGRESSION")
display(coef_df.round(4))

print("\nKoefisien model:")
for _, row in coef_df.iterrows():
    print(f"- {row['Feature']}: koefisien {row['Coefficient']:.3f}")
```

```python

# Visual validasi model: Actual vs Predicted
plt.figure(figsize=(8, 6))
sns.scatterplot(x=y_test, y=y_pred, alpha=0.45, color="#4C72B0")
min_val = min(y_test.min(), y_pred.min())
max_val = max(y_test.max(), y_pred.max())
plt.plot([min_val, max_val], [min_val, max_val], linestyle="--")
plt.title("Actual vs Predicted Performance Index")
plt.xlabel("Actual Performance Index")
plt.ylabel("Predicted Performance Index")
plt.tight_layout()

model_plot = os.path.join(OUTPUT_DIR, "06_actual_vs_predicted.png")
plt.savefig(model_plot, dpi=180, bbox_inches="tight")
plt.show()
```

```python

# Contoh prediksi beberapa siswa dari test set
prediction_sample = X_test.head(10).copy()
prediction_sample["Actual Performance Index"] = y_test.head(10).values
prediction_sample["Predicted Performance Index"] = model.predict(X_test.head(10)).round(2)
prediction_sample["Absolute Error"] = (
    prediction_sample["Actual Performance Index"] -
    prediction_sample["Predicted Performance Index"]
).abs().round(2)

display(prediction_sample)
```

### Interpretasi Model

R² menunjukkan proporsi variasi target yang dijelaskan model pada data uji; MAE dan RMSE menunjukkan besarnya kesalahan prediksi. Hasil evaluasi berlaku untuk pembagian data yang digunakan dan belum menjamin kinerja pada siswa atau periode lain.

Koefisien menggambarkan asosiasi setelah variabel lain diperhitungkan. Koefisien dan akurasi prediksi tidak membuktikan bahwa perubahan satu variabel akan menyebabkan perubahan nilai siswa.

Prediksi dapat membantu menyusun prioritas peninjauan, dengan keputusan intervensi tetap memerlukan pertimbangan pendidik.

```python
# WEEK 5 — Design System

from matplotlib.patches import Rectangle
import textwrap

WEEK5_OUTPUT_DIR = "week5_outputs"
os.makedirs(WEEK5_OUTPUT_DIR, exist_ok=True)

W5_TITLE_WRAP = 68
W5_SUBTITLE_WRAP = 88
W5_FOOTNOTE_WRAP = 105

def wrap_fig_text(text, width):
    return textwrap.fill(
        text,
        width=width,
        break_long_words=False,
        break_on_hyphens=False
    )

# Palet semantic: accent = fokus, neutral = konteks, warning = area yang perlu perhatian.
W5_ACCENT = "#1F77B4"
W5_NEUTRAL = "#B8C0CC"
W5_DARK = "#263238"
W5_MUTED = "#66727D"
W5_WARNING = "#D1495B"
W5_BG = "#FFFFFF"
W5_GRID = "#D9DEE3"

plt.rcParams.update({
    "figure.facecolor": W5_BG,
    "axes.facecolor": W5_BG,
    "axes.edgecolor": W5_BG,
    "axes.labelcolor": W5_DARK,
    "xtick.color": W5_MUTED,
    "ytick.color": W5_MUTED,
    "text.color": W5_DARK,
    "font.family": "DejaVu Sans",
    "axes.titleweight": "bold",
})

def clean_axes(ax, keep_bottom=True):
    """Minimalisasi elemen non-data tanpa mengubah data."""
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_visible(False)
    if keep_bottom:
        ax.spines["bottom"].set_color(W5_GRID)
        ax.spines["bottom"].set_linewidth(0.8)
    else:
        ax.spines["bottom"].set_visible(False)
    ax.tick_params(axis="both", length=0)

print("Week 5 design system siap.")

# Warna dibuat berdasarkan fungsi/semantic encoding
COLOR_ACCENT = "#1F77B4"
COLOR_ACCENT_DARK = "#174A73"
COLOR_NEUTRAL = "#B8C0C8"
COLOR_NEUTRAL_DARK = "#667085"
COLOR_TEXT = "#1F2937"
COLOR_SUBTEXT = "#667085"
COLOR_LIGHT = "#E8EDF2"
COLOR_WARNING = "#C0392B"
COLOR_REFERENCE = "#98A2B3"

# Typography
TITLE_SIZE = 18
SUBTITLE_SIZE = 10
AXIS_SIZE = 10
LABEL_SIZE = 10
ANNOTATION_SIZE = 10
FOOTNOTE_SIZE = 8.5

def clean_ax(ax, grid_axis=None):
    """
    Membersihkan elemen chart yang tidak diperlukan.
    """
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_visible(False)
    ax.spines["bottom"].set_color("#D0D5DD")

    ax.tick_params(axis="both", colors=COLOR_SUBTEXT, labelsize=AXIS_SIZE)

    if grid_axis == "y":
        ax.grid(axis="y", color="#E5E7EB", linewidth=0.8)
        ax.grid(axis="x", visible=False)
    elif grid_axis == "x":
        ax.grid(axis="x", color="#E5E7EB", linewidth=0.8)
        ax.grid(axis="y", visible=False)
    else:
        ax.grid(False)

    ax.set_axisbelow(True)


def add_source_note(fig, text="Source: Student Performance dataset | Exploratory analysis"):
    fig.text(
        0.01,
        0.01,
        wrap_fig_text(text, W5_FOOTNOTE_WRAP),
        ha="left",
        va="bottom",
        fontsize=FOOTNOTE_SIZE,
        color=COLOR_SUBTEXT
    )


print("Week 5 industrial design system siap.")
```

```python
# WEEK 5 — Model Prediksi / After

# Hitung residual dan error tambahan untuk analisis
prediction_eval = pd.DataFrame({
    "Actual": np.asarray(y_test),
    "Predicted": np.asarray(y_pred)
})

prediction_eval["Residual"] = (
    prediction_eval["Actual"] -
    prediction_eval["Predicted"]
)

prediction_eval["Absolute Error"] = (
    prediction_eval["Residual"].abs()
)

within_2_points = (
    prediction_eval["Absolute Error"] <= 2
).mean() * 100

mean_residual = prediction_eval["Residual"].mean()
std_residual = prediction_eval["Residual"].std()

fig, ax = plt.subplots(figsize=(9, 7))

# Scatter
ax.scatter(
    prediction_eval["Actual"],
    prediction_eval["Predicted"],
    s=24,
    alpha=0.28,
    color=COLOR_ACCENT
)

# Perfect prediction line
min_val = min(
    prediction_eval["Actual"].min(),
    prediction_eval["Predicted"].min()
)

max_val = max(
    prediction_eval["Actual"].max(),
    prediction_eval["Predicted"].max()
)

ax.plot(
    [min_val, max_val],
    [min_val, max_val],
    color=COLOR_REFERENCE,
    linewidth=2,
    linestyle="--"
)

fig.text(
    0.01,
    0.97,
    wrap_fig_text(
        f"Model mengikuti nilai aktual dengan baik pada data uji (R² = {r2:.4f})",
        W5_TITLE_WRAP
    ),
    ha="left",
    va="top",
    fontsize=TITLE_SIZE,
    fontweight="bold",
    color=COLOR_TEXT
)

fig.text(
    0.01,
    0.89,
    wrap_fig_text(
        f"MAE {mae:.2f} • RMSE {rmse:.2f} • {within_2_points:.1f}% observasi memiliki galat ≤ 2 poin",
        W5_SUBTITLE_WRAP
    ),
    ha="left",
    va="top",
    fontsize=SUBTITLE_SIZE,
    color=COLOR_SUBTEXT
)

# Reference line label
ax.text(
    max_val - 2,
    max_val - 1,
    "Prediksi sempurna",
    ha="right",
    va="top",
    fontsize=LABEL_SIZE,
    color=COLOR_SUBTEXT
)

# Model metrics block
metrics_text = (
    f"R²             {r2:.4f}\n"
    f"MAE            {mae:.2f}\n"
    f"RMSE           {rmse:.2f}\n"
    f"Rata-rata residu {mean_residual:.2f}"
)

ax.text(
    0.03,
    0.18,
    metrics_text,
    transform=ax.transAxes,
    ha="left",
    va="top",
    fontsize=10,
    color=COLOR_TEXT,
    bbox=dict(
        boxstyle="round,pad=0.5",
        facecolor="white",
        edgecolor="#D0D5DD"
    )
)

ax.set_xlabel("Performance Index Aktual", fontsize=AXIS_SIZE)
ax.set_ylabel("Performance Index Prediksi", fontsize=AXIS_SIZE)

clean_ax(ax, grid_axis="y")

fig.text(
    0.01, 0.045,
    wrap_fig_text(
        "Rekomendasi aksi: gunakan model sebagai alat prioritas untuk menandai siswa yang perlu ditinjau, bukan sebagai satu-satunya dasar keputusan intervensi akademik.",
        W5_FOOTNOTE_WRAP
    ),
    ha="left", va="bottom", fontsize=10, color=COLOR_TEXT
)

add_source_note(
    fig,
    "Sumber: Linear Regression | pembagian data uji random_state=42 | metrik berlaku pada data uji ini"
)

plt.tight_layout(rect=[0, 0.085, 1, 0.86])

model_makeover_path = os.path.join(
    WEEK5_OUTPUT_DIR,
    "W5_06_model_actual_vs_predicted_makeover.png"
)

plt.savefig(
    model_makeover_path,
    dpi=220,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()

print("Model makeover saved:", model_makeover_path)
print(f"R²                 : {r2:.4f}")
print(f"MAE                : {mae:.4f}")
print(f"RMSE               : {rmse:.4f}")
print(f"Error <= 2 points  : {within_2_points:.2f}%")
print(f"Mean residual      : {mean_residual:.4f}")
print(f"Residual std       : {std_residual:.4f}")
```
